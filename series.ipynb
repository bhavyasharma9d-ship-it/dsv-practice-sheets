```python

sales = pd.Series(
    [450, 650, 800, 300],
    index=['Laptop', 'Phone', 'Tablet', 'Mouse']
)


print("Values:", sales.values)

print("Index:", sales.index)


print("Phone sales:", sales['Phone'])


print("Sales above 500:")
print(sales[sales > 500])


print("Is Laptop present?", 'Laptop' in sales.index)


print("Dictionary:", sales.to_dict())
```
